```python
from langchain_text_splitters import MarkdownHeaderTextSplitter
```

```python
import os
print(os.listdir('/content/drive/MyDrive/OOP'))
```

```python
with open('/home/hungmanh/Documents/CodeMentor/app/data/bai2.txt', 'r', encoding='utf-8') as f:
    text = f.read()
print(text)
```

```python
docs = [text]
splitter = MarkdownHeaderTextSplitter(
        headers_to_split_on=[
            ("#", "Header_1"),
            ("##", "Header_2"),
        ],
    )
splits = [split for doc in docs for split in splitter.split_text(doc)]
```

```python
for split in splits:
    print("\n--- Split ---\n\n")
    print(split.page_content)
```

```python
output_path = '/home/hungmanh/Documents/CodeMentor/app/data/splits_output.txt'
with open(output_path, 'w', encoding='utf-8') as f:
    for split in splits:
        f.write(str(split))
        f.write('\n\n--- Split ---\n\n')
print(f"Đã lưu các splits vào {output_path}")
```
